# ЗАДАЧА 1 (ABBREVIATION)
Составьте аббревиатуру из первых букв слов.\
Однобуквенные слова нужно пропустить независимо от регистра.\
Результат выведите прописными буквами.\
Пример: 'Белорусский государственный университет информатики и радиоэлектроники' → 'БГУИР'.\
phrase = input("Введите фразу, из которой нужно сделать аббревиатуру:\n")\

```python
phrase = input("Введите фразу, из которой нужно сделать аббревиатуру:\n")
result = ""
for word in phrase.split():
    if len(word) > 1:
        result += word[0].upper()
print(result)
```

# ЗАДАЧА 2 (CALCULATOR)
Напишите программу, которое получает математическое выражение и вычисляет ее результат.\
В выражении могут быть либо только операции суммирования, либо только операции умножения:\
Пример: для строки `1+2+3+4` программа должна посчитать 10.\
Пример: для строки `10*20*30*40` программа должна посчитать 240000.\
expression = input("Введите выражение (сумму или произведение): ")\

```python
expression = input("Введите выражение (сумму или произведение): ")

if "+" in expression:
    numbers = expression.split("+")
    result = 0
    for num in numbers:
        result += int(num)

elif "*" in expression:
    numbers = expression.split("*")
    result = 1
    for num in numbers:
        result *= int(num)

print(result)
```
