# ЗАДАЧА 1 (ABBREVIATION)

```python
phrase = input("Введите фразу, из которой нужно сделать аббревиатуру:\n")
result = ""
for word in phrase.split():
    if len(word) > 1:
        result += word[0].upper()
print(result)
```

# ЗАДАЧА 2 (CALCULATOR)

```python
expression = input("Введите выражение (сумму или произведение): ")

if "+" in expression:
    numbers = expression.split("+")
    result = 0
    for num in numbers:
        result += int(num)

elif "*" in expression:
    numbers = expression.split("*")
    result = 1
    for num in numbers:
        result *= int(num)

print(result)
```

# ЗАДАЧА 3 (CHESS_CASE)

```python
sentence = input("Введите предложение: ")

result = ""

is_upper = True

for word in sentence.split():
    if is_upper:
        result += word.upper() + " "
    else:
        result += word.lower() + " "

    # Меняем значение на противоположное (True станет False, а False станет True)
    is_upper = not is_upper

# Удаляем один лишний пробел, который прибавился в самом конце
result = result.rstrip()

print(result)
```

# ЗАДАЧА 4 ()

```python
text = input("Введите текст: ")

result = ""

for word in text.split():
    # Очищаем слово от знаков препинания по краям (запятые, точки, восклицательные знаки)
    clean_word = word.strip(".,!?\"';:()[]{}")
    
    if clean_word.startswith("#") and len(clean_word) > 1:
        if clean_word not in result:
            result += clean_word + " "

# Удаляем лишний пробел, который добавился в самом конце
result = result.rstrip()

print(result)
```
