# ЗАДАЧА 1 (ABBREVIATION)

```python
phrase = input("Введите фразу, из которой нужно сделать аббревиатуру:\n")
result = ""
for word in phrase.split():
    if len(word) > 1:
        result += word[0].upper()
print(result)
```

# ЗАДАЧА 2 (CALCULATOR)

```python
expression = input("Введите выражение (сумму или произведение): ")

if "+" in expression:
    numbers = expression.split("+")
    result = 0
    for num in numbers:
        result += int(num)

elif "*" in expression:
    numbers = expression.split("*")
    result = 1
    for num in numbers:
        result *= int(num)

print(result)
```
